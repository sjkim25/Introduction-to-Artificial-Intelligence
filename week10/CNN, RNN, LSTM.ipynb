**CNN (Convolutional Neural Network, 합성곱 신경망)**

이미지의 공간적 정보를 보존하고 효율적으로 특징을 추출하는 모델


1.   데이터 처리 방식의 변화
*   평탄화의 한계: 데이터를 1차원으로 펼치면 인접한 픽셀 간의 공간적 특징이 소실된다
*   합성곱 연산: 원본 데이터 위에 특정 크기의 필터를 겹쳐 곱셈 연산을 수행, 공간적 구조를 유지한 채 새로윤 특징 맵 생성

2.   주요 구성 요소 및 최적화
*   필터와 가중치: 필터는 학습을 통해 최적화되는 Weight와 Bias의 집합
*   Stride와 Padding: Stride는 필터가 이동하는 간격 결정, Padding은 외각에 값을 채우는 것

3.   정보 요약 및 불변성 확보
*   최대 풀링(Max pooling): 특정 영역에서 가장 큰 값을 선택
*   평균 풀링(Average Pooling): 특정 영역에서 평균값을 산출
*   이동 불변성: 풀링을 통해 특징의 위치가 미세하게 변하더라도 동일한 특징으로 인식 가능하게 하는 능력

**RNN (Recurrent Neural Network, 순환 신경망)**

데이터의 순서와 시간적 흐름을 학습에 반영하는 모델


*   순환 구조: 은닉층의 출력값이 다시 다음 시점의 입력값과 함께 사용되는 재귀적 구조
*   가중치 공유: 모든 시점에서 동일한 필터를 반복적으로 사용, 데이터 길이에 관계없이 고정된 수의 파라미터로 시계열 데이터 처리 가능

**LSTM (Long Short-Term Memory, 장단기 메모리)**

RNN의 장기 의존성 결여 문제를 해결하기 위해 설계된 순환 구조 모델


*   셀 상태(Cell State): 정보 손실 없이 먼 시점까지 전달
*   게이트 시스템:

1.   망각 게이트(Forget Gate): 과거 정보 중 불필요한 것 선별 삭제
2.   입력 게이트(Input Gate): 현시점 들어온 정보 중 가치 있는 것을 셀 상태에 저장
3. 출력 게이트(Output Gate): 업데이트된 셀 상태를 기반으로 최종 출력값 결정
